```python
## Activate the packages required
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
## Load the data
df = pd.read_csv("insurance.csv")
df.head()
```

```python
# check the number of rows & columns
df.shape
```

```python
# check for missing values
df.isnull().sum()
```

```python
# convert all categorical variables (sex, smoker, region) into dummy nominal variables
df['sex'] = df['sex'].apply({'male':0, 'female':1}.get)
df['smoker'] = df['smoker'].apply({'yes':1, 'no':0}.get)
df['region'] = df['region'].apply({'southwest':1, 'southeast':2,'northwest':3,'northeast':4}.get)
df.head()
```

```python
# Define the IVs & DV (STEP-1 in Machine Learning Modeling)
x = df.drop(['charges'], axis=1)   ## except charges, all other variables are IVs
y = df.charges   # charges is the DV
```

```python
# Split the data into Training & Test Data (Step-2 in ML Modelling)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state=42)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
# Define the ML Model (Step-3 in ML Modelling)
model = LinearRegression()
```

```python
# Fit the model (Step-4 in ML Modelling)
model.fit(x_train, y_train)  # estimate the alpha and betas
```

```python
print(model.intercept_, model.coef_, model.score(x,y))
```

```python
# Predict using the model for test data (Step-5 of ML Modelling)
predictions = model.predict(x_test)
```

```python
# Measure the accuracy using suitable metric (Step-6 of ML Modelling)
print("R2 score:", r2_score(y_test, predictions))
```

```python
# User input data for prediction using the above model
new_cust_data = {'age': 30, 'sex':1, 'bmi': 30, 'children': 4, 'smoker': 1, 'region':1}
index = [0]
new_cust_data_df = pd.DataFrame(new_cust_data, index)
new_cust_data_df
```

```python
# Prediction
predicted_charges = model.predict(new_cust_data_df)
print("The charges for this customer are:", predicted_charges)
```
