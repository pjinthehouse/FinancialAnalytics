```python
# Ml Based stock price prediction & algo trading backtesting
# Importing the required libraries
import yfinance as yf
import numpy as np

data = yf.download('NESTLEIND.NS', start='2020-01-01')
data['returns'] = np.log(data.Close.pct_change()+1)
```

```python
# create a function to create lag returns based on the given return data
def lagit(data, lags):
    names = []
    for i in range(1, lags + 1):
        data['Lag_'+str(i)] = data['returns'].shift(i)
        names.append('Lag_'+str(i))
    return names

# create lagnames of your choice (demo: 5 day lags)
lagnames = lagit(data,5)
lagnames
```

```python
## Drop missing values
data.dropna(inplace=True)
```

```python
## print the dataframe (entire data)
data
```

```python
## Step-1: Define the X and Y variables
X = data[lagnames] 
Y = data['returns']
```

```python
## Step-2: Split the data into TRAINING & TEST DATA
# import dependencies for machine learning data analysis
from sklearn.model_selection import train_test_split
## split the data into training and test data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, shuffle=False, test_size = 0.3, random_state = 101)
X.shape, Y.shape, X_train.shape, X_test.shape, Y_train.shape,  Y_test.shape
```

```python
## Step-3: Define your regression model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
## Step-4: Fit the model / Estimate the parameters (a & bs) using the TRAINING data
model.fit(X_train,Y_train)
model.coef_
```
