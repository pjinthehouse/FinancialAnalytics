```python
def calculate_future_value(present_value, inflation_rate, years):
    future_value = present_value * ((1 + inflation_rate) ** years)
    return future_value

def calculate_monthly_savings(shortfall, expected_return, years):
    months = years * 12
    monthly_savings = (shortfall / ((((1+expected_return/12)**months)-1) / 
                                    (expected_return/12)))*(1/(1+expected_return/12))
    return monthly_savings

def main():
    current_age = int(input("Enter current age of the child: "))
    milestone_age = int(input("Enter milestone age when corpus amount is needed: "))
    corpus_amount = float(input("Enter corpus amount needed at today's cost: "))
    inflation_rate = float(input("Enter expected inflation rate (in decimal): "))
    expected_return = float(input("Enter expected return on savings (in decimal): "))
    existing_savings = float(input("Enter existing savings if any: "))

    years_until_milestone = milestone_age - current_age

    future_value_corpus = calculate_future_value(corpus_amount, inflation_rate, years_until_milestone)
    future_value_savings = calculate_future_value(existing_savings, expected_return, years_until_milestone)
    shortfall = future_value_corpus - future_value_savings
    monthly_savings_required = calculate_monthly_savings(shortfall, expected_return, years_until_milestone)

    print(f"\nFuture value of corpus amount needed: {future_value_corpus:.2f}")
    print(f"Future value of existing savings: {future_value_savings:.2f}")
    print(f"Shortfall: {shortfall:.2f}")
    print(f"Monthly savings required: {monthly_savings_required:.2f}")

if __name__ == "__main__":
    main()
```
